# Get model predictions based on parameters

```python
import numpy as np
import pandas as pd

from enum import IntEnum

from model_utils import *
```

```python
params = {
    "alpha": 1,
    "w_r": 1,
    "w_c": 1,
    "w_e": 1,
}

output_mappings = {
    "ZERO": 0,
    "FIFTY": 50,
    "SEVENTY_FIVE": 75,
    "ONE_HUNDRED": 100,
    "LOW": "low",
    "HIGH": "high"
}
```

```python
def predict_actor_all(params):
    result_discomfort_only = actor_discrete_discomfort_only(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )
    result_vanilla_inv_plan = actor_discrete_vanilla_inv_plan(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )
    result_full_model = actor_discrete_full_model(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )

    result_mappings = {
        "discomfort_only": result_discomfort_only,
        "vanilla_inv_plan": result_vanilla_inv_plan,
        "full_model": result_full_model,
    }

    dfs = []
    for model_label, result in result_mappings.items():
        df = result[1].pandas
        df = df.rename(
            columns={
                "reward_condition": "reward",
                "relationship_condition": "intimacy",
                df.columns[-1]: "p_action",
            }
        ).replace(output_mappings)
        df["model"] = model_label
        dfs.append(df)

    df_combined = pd.concat(dfs)

    df_combined.insert(0, "w_c", params["w_c"])
    df_combined.insert(0, "w_r", params["w_r"])
    df_combined.insert(0, "w_e", params["w_e"])
    df_combined.insert(0, "alpha", params["alpha"])
    return df_combined


def predict_observer_intimacy_all(params):
    result_discomfort_only = observer_intimacy_discomfort_only(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )
    result_vanilla_inv_plan = observer_intimacy_vanilla_inv_plan(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )
    result_full_model = observer_intimacy_full_model(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )

    result_mappings = {
        "discomfort_only": result_discomfort_only,
        "vanilla_inv_plan": result_vanilla_inv_plan,
        "full_model": result_full_model,
    }

    dfs = []
    for model_label, result in result_mappings.items():
        df = result[1].pandas
        df = df.rename(
            columns={
                "reward_condition": "reward",
                "relationship": "intimacy",
                df.columns[-1]: "density",
            }
        ).replace(output_mappings)
        df["model"] = model_label
        dfs.append(df)

    df_combined = pd.concat(dfs)

    df_combined.insert(0, "w_c", params["w_c"])
    df_combined.insert(0, "w_r", params["w_r"])
    df_combined.insert(0, "w_e", params["w_e"])
    df_combined.insert(0, "alpha", params["alpha"])

    return df_combined


def predict_observer_reward_all(params):
    result_discomfort_only = observer_reward_discomfort_only(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )
    result_vanilla_inv_plan = observer_reward_vanilla_inv_plan(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )
    result_full_model = observer_reward_full_model(
        alpha=params["alpha"],
        w_r=params["w_r"],
        w_c=params["w_c"],
        w_e=params["w_e"],
        return_pandas=True,
    )

    result_mappings = {
        "discomfort_only": result_discomfort_only,
        "vanilla_inv_plan": result_vanilla_inv_plan,
        "full_model": result_full_model,
    }

    dfs = []
    for model_label, result in result_mappings.items():
        df = result[1].pandas
        df = df.rename(
            columns={
                "reward_condition": "reward",
                "relationship_condition": "intimacy",
                df.columns[-1]: "p_reward",
            }
        ).replace(output_mappings)
        df["model"] = model_label
        dfs.append(df)

    df_combined = pd.concat(dfs)

    df_combined.insert(0, "w_c", params["w_c"])
    df_combined.insert(0, "w_r", params["w_r"])
    df_combined.insert(0, "w_e", params["w_e"])
    df_combined.insert(0, "alpha", params["alpha"])

    return df_combined
```

```python
actor_results = predict_actor_all(params)
observer_intimacy_results = predict_observer_intimacy_all(params)
observer_reward_results = predict_observer_reward_all(params)
```

```python
# different params for each model, for now, just for visualization
actor_params = {
    "alpha": 1,
    "w_r": 1,
    "w_c": 1,
    "w_e": 1,
}
observer_intimacy_params = {
    "alpha": 2,
    "w_r": 2,
    "w_c": 1,
    "w_e": 1,
}
observer_reward_params = {
    "alpha": 2,
    "w_r": 2,
    "w_c": 1,
    "w_e": 1,
}
```

```python
def predict_and_output_all(actor_params, observer_intimacy_params, observer_reward_params):
    actor_results = predict_actor_all(actor_params)
    actor_results.to_csv(
        f"actor_preds.csv",
        index=False,
    )
    observer_intimacy_results = predict_observer_intimacy_all(observer_intimacy_params)
    observer_intimacy_results.to_csv(
        f"observer_intimacy_preds.csv",
        index=False,
    )
    observer_reward_results = predict_observer_reward_all(observer_reward_params)
    observer_reward_results.to_csv(
        f"observer_reward_preds.csv",
        index=False,
    )
    return None
```
